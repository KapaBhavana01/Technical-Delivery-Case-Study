```python
import pandas as pd
def _is_true(c: pd.Series) -> pd.Series:
    return c == "t"
```

```python
companies["iata_approved"] = _is_true(companies["iata_approved"])
companies.head()
```

```python
def parse_percent(c: pd.Series) -> pd.Series:
    return c.str.replace("%","").astype(float)
```

```python
companies["company_rating"]= parse_percent(companies["company_rating"])
companies.head()
```

```python
shuttles = catalog.load("shuttles")
shuttles.head()
```

```python
shuttles.dtypes
```

```python
def parse_money(c: pd.Series) -> pd.Series:
    return c.str.replace("$","").str.replace(",","").astype(float)
```

```python
shuttles["d_check_complete"] = _is_true(shuttles["d_check_complete"])
shuttles["moon_clearance_complete"] = _is_true(shuttles["moon_clearance_complete"])
```

```python
shuttles["price"] = parse_money(shuttles["price"])
```

```python
shuttles.head()
```

```python
reviews = catalog.load("reviews")
```

```python
reviews.head()
```

```python
rated_shuttles = shuttles.merge(reviews, left_on = "id", right_on = "shuttle_id")
rated_shuttles.head()
```

```python
model_input_table = rated_shuttles.merge(companies, left_on = "company_id", right_on = "id")
model_input_table.head()
```
